```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.stats.multicomp import pairwise_tukeyhsd
from openpyxl import load_workbook

# Configurazione stile grafici per pubblicazione
sns.set_theme(style="ticks")
plt.rcParams.update({
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 13,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'figure.titlesize': 14
})

# Palette cromatica fissa richiesta dal vincolo (Gradiente di Blu)
PALETTE_CONDIZIONI = {
    'A': '#0B3C5D',  # Dark Blue (Actor)
    'O': '#328CC1',  # Medium Blue (Observer)
    'C': '#D9B310' if False else '#98D7C2' # Light Blue/Teal (Control)
}
# Per un gradiente blu puro e coerente:
PALETTE_CONDIZIONI = {'A': '#104E8B', 'O': '#4682B4', 'C': '#B0E2FF'}

# Costanti sperimentali
COMPORTAMENTI = ['headshaking', 'beakwiping', 'preening', 'scratching']
TOTAL_FRAMES_VIDEO = 5000  # Normalizzazione richiesta a 5000fps/frames totali

# ==========================================
# BLOCK 1: RAW DATA LOADING & VALIDATION
# ==========================================

def load_and_validate_data(file_path="imitation_data.xlsx"):
    """
    Carica il file Excel e valida la presenza della tabella metadati 'frames_behav'.
    Verifica che non manchino colonne cruciali.
    """
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"Errore: Il file '{file_path}' non è stato trovato nella cartella corrente.")
    
    try:
        # Carica foglio metadati
        metadata = pd.read_excel(file_path, sheet_name="frames_behav")
    except ValueError:
        raise ValueError("Errore: Il foglio 'frames_behav' è assente nel file Excel.")
        
    # Controllo colonne richieste
    colonne_richieste = ['SubID', 'Condition', 'Sex']
    for col in colonne_richieste:
        if col not in metadata.columns:
            raise ValueError(f"Errore: La colonna obbligatoria '{col}' è assente nel foglio 'frames_behav'.")
            
    print(f"[INFO] Metadati caricati correttamente. Rilevati {len(metadata)} soggetti.")
    return metadata

# ==========================================
# BLOCK 2: BEHAVIORAL EVENT EXTRACTION (+1)
# ==========================================

def calcola_occorrenze_soggetto(df_soggetto):
    """
    Calcola le occorrenze discrete basate sui cambi di stato (differenza row-to-row == +1).
    Gestisce l'eccezione del frame index 0.
    """
    conteggi = {}
    for comp in COMPORTAMENTI:
        if comp not in df_soggetto.columns:
            conteggi[comp] = 0
            continue
            
        serie = df_soggetto[comp].to_numpy()
        if len(serie) == 0:
            conteggi[comp] = 0
            continue
            
        # Calcolo transizioni (+1 indica passaggio da 0 a 1)
        differenze = np.diff(serie)
        transizioni_positive = np.sum(differenze == 1)
        
        # Regola d'eccezione per il frame iniziale
        if serie[0] == 1:
            transizioni_positive += 1
            
        conteggi[comp] = transizioni_positive
    return conteggi

def estrai_e_salva_occorrenze(file_path="imitation_data.xlsx", metadata=None):
    """
    Cicla su tutti i fogli dei soggetti, estrae le occorrenze e i frame totali spesi,
    generando il foglio riassuntivo 'n_behav'.
    """
    if metadata is None:
        metadata = load_and_validate_data(file_path)
        
    xl = pd.ExcelFile(file_path)
    lista_fogli = xl.sheet_names
    
    righe_occorrenze = []
    righe_durata_frames = []
    
    for _, row in metadata.iterrows():
        sub_id = str(row['SubID'])
        cond = row['Condition']
        sesso = row['Sex']
        
        if sub_id in lista_fogli:
            df_sub = pd.read_excel(file_path, sheet_name=sub_id)
            
            # 1. Calcolo Occorrenze discrete
            occorrenze = calcola_occorrenze_soggetto(df_sub)
            riga_occ = {'SubID': sub_id, 'Condition': cond, 'Sex': sesso}
            riga_occ.update(occorrenze)
            righe_occorrenze.append(riga_occ)
            
            # 2. Calcolo Somma totale dei Frame (per la normalizzazione successiva)
            riga_durata = {'SubID': sub_id, 'Condition': cond, 'Sex': sesso}
            for comp in COMPORTAMENTI:
                riga_durata[comp] = int(df_sub[comp].sum()) if comp in df_sub.columns else 0
            righe_durata_frames.append(riga_durata)
        else:
            print(f"[AVVISO] Soggetto {sub_id} presente nei metadati ma privo di foglio dati dedicato.")

    df_n_behav = pd.DataFrame(righe_occorrenze)
    df_frames_somma = pd.DataFrame(righe_durata_frames)
    
    # Salvataggio nel file Excel senza sovrascrivere i fogli esistenti
    book = load_workbook(file_path)
    # Rimuovi il foglio n_behav se già esisteva per evitare duplicati incrementali
    if "n_behav" in book.sheetnames:
        del book["n_behav"]
    book.save(file_path)
    book.close()
    
    # Scrittura pulita del nuovo foglio
    with pd.ExcelWriter(file_path, engine="openpyxl", mode="a", if_sheet_exists="replace") as writer:
        df_n_behav.to_excel(writer, sheet_name="n_behav", index=False)
        
    print("[INFO] Estrazione completata. Creato il foglio 'n_behav' all'interno dell'Excel.")
    return df_n_behav, df_frames_somma

# ==========================================
# BLOCK 3: DESCRIPTIVE STATISTICAL ANALYSIS
# ==========================================

def esegui_descrittive(df_analisi):
    """
    Calcola e stampa le tabelle descrittive raggruppate per Condizione e Sesso.
    """
    print("\n" + "="*60)
    print("          BLOCK 3: STATISTICHE DESCRITTIVE")
    print("="*60)
    
    for comp in COMPORTAMENTI:
        print(f"\n--- Comportamento: {comp.upper()} ---")
        desc = df_analisi.groupby(['Condition', 'Sex'])[comp].agg(['count', 'mean', 'median', 'std'])
        print(desc.round(3))

# ==========================================
# BLOCK 4: INFERENTIAL STATISTICAL EVALUATIONS
# ==========================================
def esegui_inferenziale(df_analisi):
    """
    Esegue ANOVA a una via tra le condizioni, test Post-Hoc Tukey HSD 
    e controlla gli effetti principali/interazioni del sesso.
    Previene crash in caso di varianza zero.
    """
    import statsmodels.formula.api as smf  # <--- CORREZIONE QUI
    
    print("\n" + "="*60)
    print("          BLOCK 4: STATISTICA INFERENZIALE")
    print("="*60)
    
    risultati_tukey = {}
    
    for comp in COMPORTAMENTI:
        print(f"\n\n##### ANALISI PER: {comp.upper()} #####")
        
        # Controllo varianza interna ai gruppi per evitare crash statistici
        gruppi = [df_analisi[df_analisi['Condition'] == c][comp].values for c in ['A', 'O', 'C']]
        varianze = [np.var(g) for g in gruppi]
        
        if sum(varianze) == 0:
            print(f"[SKIP] {comp} ha varianza zero in tutti i gruppi. Impossibile calcolare ANOVA.")
            continue
            
        # 1. ANOVA a una via (Effetto Condizione)
        f_stat, p_val = stats.f_oneway(*gruppi)
        print(f"ANOVA a una via (Condizione A, O, C): F = {f_stat:.3f}, p-value = {p_val:.5f}")
        
        # 2. Post-hoc Tukey HSD se l'ANOVA globale è significativa
        if p_val < 0.05:
            print("-> ANOVA Significativa (p < 0.05). Avvio Test Post-Hoc Tukey HSD...")
            tukey = pairwise_tukeyhsd(endog=df_analisi[comp], groups=df_analisi['Condition'], alpha=0.05)
            print(tukey)
            risultati_tukey[comp] = tukey
        else:
            print("-> ANOVA non significativa. I gruppi non mostrano divergenze globali.")
            
        # 3. Effetti di Sesso e Interazioni tramite Modello Lineare (ANOVA a 2 vie)
        try:
            # Usiamo smf.ols invece di sm.Formula.ols
            modello = smf.ols(f"{comp} ~ C(Condition) * C(Sex)", data=df_analisi).fit()
            tabella_anova2 = sm.stats.anova_lm(modello, typ=2)
            print("\nANOVA a Due Vie (Fattori: Condizione, Sesso, Interazione):")
            print(tabella_anova2.round(5))
        except Exception as e:
            print(f"[AVVISO] Impossibile stimare l'interazione del sesso per {comp}: {e}")
            
    return risultati_tukey

    
# ==========================================
# BLOCK 5: DATA VISUALIZATION
# ==========================================
def annota_significativita(ax, tukey_res, comportamento, df_dati, ordine_gruppi=['A', 'O', 'C']):
    """
    Disegna le linee di significatività estraendo i dati in modo sicuro e pulito,
    eliminando gli spazi bianchi che causano il mancato riconoscimento dei gruppi.
    """
    if not tukey_res or comportamento not in tukey_res:
        return
        
    # Otteniamo i dati strutturati direttamente dall'oggetto Tukey senza rischiare con stringhe grezze
    tukey_data = tukey_res[comportamento]
    
    # Mappa delle coordinate x dei gruppi sul grafico
    x_coords = {gruppo: idx for idx, gruppo in enumerate(ordine_gruppi)}
    
    # Calcolo dinamico delle altezze delle barre basato sui dati reali del singolo quadrante
    y_min, y_max = ax.get_ylim()
    spostamento_y = (y_max - y_min) * 0.06
    # CORREZIONE: Usiamo 'df_dati' (passato come argomento) invece del vecchio 'df_occorrenze' globale
    livello_linea = df_dati[comportamento].max() + (spostamento_y * 0.5)

    # Cicliamo sui risultati interni del test di Tukey
    for i, g1 in enumerate(tukey_data.groupsunique):
        for j, g2 in enumerate(tukey_data.groupsunique):
            if i >= j:
                continue
                
            g1_clean = str(g1).strip()
            g2_clean = str(g2).strip()
            
            for row in tukey_data.summary().data[1:]:
                row_g1 = str(row[0]).strip()
                row_g2 = str(row[1]).strip()
                
                if (row_g1 == g1_clean and row_g2 == g2_clean) or (row_g1 == g2_clean and row_g2 == g1_clean):
                    p_val = row[5] # Il p-adj si trova sempre all'indice 5
                    
                    if p_val < 0.05:
                        if p_val < 0.001: asterischi = "***"
                        elif p_val < 0.01: asterischi = "**"
                        else: asterischi = "*"
                        
                        x1 = x_coords.get(g1_clean)
                        x2 = x_coords.get(g2_clean)
                        
                        if x1 is not None and x2 is not None:
                            ax.plot([x1, x1, x2, x2], 
                                    [livello_linea, livello_linea + (spostamento_y * 0.3), livello_linea + (spostamento_y * 0.3), livello_linea], 
                                    lw=1.1, c='black')
                            ax.text((x1 + x2) * 0.5, livello_linea + (spostamento_y * 0.4), asterischi, 
                                    ha='center', va='bottom', color='black', weight='bold', fontsize=11)
                            
                            livello_linea += spostamento_y * 1.3

def genera_grafici_quadranti(df_dati, tipo_misura="Occorrenze", tukey_res=None):
    """
    Genera una figura a 4 quadranti (2x2) per i quattro comportamenti target.
    Aggiunge le distribuzioni dei punti reali (stripplot).
    """
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    axes = axes.flatten()
    
    for idx, comp in enumerate(COMPORTAMENTI):
        ax = axes[idx]
        
        sns.boxplot(
            data=df_dati, x='Condition', y=comp, order=['A', 'O', 'C'],
            hue='Condition', palette=PALETTE_CONDIZIONI, legend=False, ax=ax, width=0.5, fliersize=0
        )
        
        sns.stripplot(
            data=df_dati, x='Condition', y=comp, order=['A', 'O', 'C'],
            color='black', alpha=0.4, size=4, jitter=0.15, ax=ax
        )
        
        ax.set_title(f"Pattern: {comp.upper()}")
        ax.set_xlabel("Condizione Sperimentale")
        ax.set_ylabel(f"Frequenza ({tipo_misura})")
        
        if tipo_misura == "Occorrenze" and tukey_res:
            # CORREZIONE: Passiamo anche 'df_dati' come quarto argomento a questa funzione
            annota_significativita(ax, tukey_res, comp, df_dati)
            
    plt.suptitle(f"Analisi Comportamentale Comparativa nei Pulcini ({tipo_misura})", y=0.98)
    plt.tight_layout()
    plt.show()

def genera_grafico_sesso(df_dati):
    """
    Genera i grafici suddivisi e categorizzati per la variabile Sex (M vs F).
    """
    fig, axes = plt.subplots(2, 2, figsize=(13, 11))
    axes = axes.flatten()
    
    for idx, comp in enumerate(COMPORTAMENTI):
        ax = axes[idx]
        
        sns.boxplot(
            data=df_dati, x='Condition', y=comp, hue='Sex', order=['A', 'O', 'C'],
            palette=['#4F94CD', '#FFB6C1'], ax=ax, width=0.6
        )
        sns.stripplot(
            data=df_dati, x='Condition', y=comp, hue='Sex', order=['A', 'O', 'C'],
            dodge=True, color='black', alpha=0.3, size=3.5, ax=ax
        )
        
        ax.set_title(f"Effetto Sesso su: {comp.upper()}")
        ax.set_xlabel("Condizione")
        ax.set_ylabel("Occorrenze")
        
        # Pulisce le legende doppie generate da stripplot
        handles, labels = ax.get_legend_handles_labels()
        ax.legend(handles[0:2], labels[0:2], title="Sesso")
        
    plt.suptitle("Divergenze Comportamentali condizionate dal Dimorfismo Sessuale", y=0.98)
    plt.tight_layout()
    plt.show()

# ==========================================
# FUNZIONE PRINCIPALE: PIPELINE RUNNER
# ==========================================

def run_full_pipeline():
    print("[PIPELINE] Avvio del processo di analisi dei dati...")
    
    # Blocco 1: Caricamento
    metadata = load_and_validate_data("imitation_data.xlsx")
    
    # Blocco 2: Calcolo transizioni di stato discrete (+1)
    df_occorrenze, df_frames_totali = estrai_e_salva_occorrenze("imitation_data.xlsx", metadata)
    
    # Blocco 3: Calcolo delle descrittive
    esegui_descrittive(df_occorrenze)
    
    # Blocco 4: Elaborazione statistica inferenziale (ANOVA e Tukey HSD)
    risultati_statistici_tukey = esegui_inferenziale(df_occorrenze)
    
    # Preparazione dati normalizzati (Normalizzazione richiesta basata su 5000 frames)
    df_normalizzato = df_frames_totali.copy()
    for comp in COMPORTAMENTI:
        df_normalizzato[comp] = df_normalizzato[comp] / TOTAL_FRAMES_VIDEO
        
    # Blocco 5: Generazione dei set grafici pubblicabili
    print("\n[INFO] Generazione Figure 1: Boxplots Occorrenze Assolute...")
    genera_grafici_quadranti(df_occorrenze, tipo_misura="Occorrenze", tukey_res=risultati_statistici_tukey)
    
    print("\n[INFO] Generazione Figure 2: Boxplots Frazione di Frame Spesi (Normalizzati)...")
    genera_grafici_quadranti(df_normalizzato, tipo_misura="Frazione Frame Normalizzati", tukey_res=None)
    
    print("\n[INFO] Generazione Figure 3: Boxplots Comparativi divisi per Sesso...")
    genera_graphic_sesso = genera_grafico_sesso(df_occorrenze)
    
    print("\n[PIPELINE EXECUTED] Pipeline completata con successo! Tutti i dati e i grafici sono stati aggiornati.")

# Avvio del flusso di lavoro completo
run_full_pipeline()
```
